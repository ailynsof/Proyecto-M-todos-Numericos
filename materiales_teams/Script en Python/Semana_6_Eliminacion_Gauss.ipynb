## Sistema de ejemplo

Consideremos el sistema

$$
\begin{aligned}
2x_1+x_2-3x_3 &= 5,\\
4x_1+3x_2-5x_3 &= 9,\\
-2x_1+2x_2+4x_3 &= 7.
\end{aligned}
$$

Por tanto,

$$
A=
\begin{bmatrix}
2&1&-3\\
4&3&-5\\
-2&2&4
\end{bmatrix},
\qquad
b=
\begin{bmatrix}
5\\9\\7
\end{bmatrix}.
$$

En Python ingresamos la matriz como una lista de filas y el vector como una lista.

````python
A = [
    [2, 1, -3],
    [4, 3, -5],
    [-2, 2, 4]
]

b = [5, 9, 7]

print("A =")
for fila in A:
    print(fila)

print("\nb =", b)
````

# 1. Eliminación de la primera incógnita

Este algoritmo elimina únicamente $x_1$ de todas las ecuaciones posteriores a la primera.

El multiplicador utilizado es

$$
m=\frac{a_{j1}}{a_{11}}.
$$

Después se actualizan los coeficientes mediante

$$
a'_{jk}=a_{jk}-ma_{1k},
$$

y el lado derecho mediante

$$
b'_j=b_j-mb_1.
$$

````python
def eliminar_primera_incognita(A, b):
    n = len(A)

    for j in range(1, n):

        m = A[j][0] / A[0][0]
        A[j][0] = 0

        for k in range(1, n):
            A[j][k] = A[j][k] - m * A[0][k]

        b[j] = b[j] - m * b[0]

    return A, b
````

### Aplicación al sistema de ejemplo

Se hace una copia de $A$ y $b$ para conservar los datos originales.

````python
from copy import deepcopy

A1 = deepcopy(A)
b1 = b.copy()

A1, b1 = eliminar_primera_incognita(A1, b1)

print("Matriz después de eliminar x1:")
for fila in A1:
    print(fila)

print("\nVector b modificado:")
print(b1)
````

El resultado esperado es

$$
\begin{bmatrix}
2&1&-3\\
0&1&1\\
0&3&1
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
=
\begin{bmatrix}
5\\-1\\12
\end{bmatrix}.
$$

# 2. Eliminación hacia adelante de Gauss

El segundo algoritmo generaliza el procedimiento anterior.

Ahora el pivote cambia en cada etapa:

$$
a_{11},\;a_{22},\;a_{33},\;\ldots
$$

El multiplicador es

$$
m=\frac{a_{ji}}{a_{ii}}.
$$

El proceso continúa hasta obtener una matriz triangular superior.

````python
def eliminacion_gauss(A, b):
    n = len(A)

    for i in range(n - 1):

        for j in range(i + 1, n):

            m = A[j][i] / A[i][i]
            A[j][i] = 0

            for k in range(i + 1, n):
                A[j][k] = A[j][k] - m * A[i][k]

            b[j] = b[j] - m * b[i]

    return A, b
````

### Aplicación al sistema de ejemplo

````python
A2 = deepcopy(A)
b2 = b.copy()

A2, b2 = eliminacion_gauss(A2, b2)

print("Matriz triangular superior:")
for fila in A2:
    print(fila)

print("\nVector b modificado:")
print(b2)
````

El resultado esperado es

$$
\begin{bmatrix}
2&1&-3\\
0&1&1\\
0&0&-2
\end{bmatrix}
\begin{bmatrix}
x_1\\x_2\\x_3
\end{bmatrix}
=
\begin{bmatrix}
5\\-1\\15
\end{bmatrix}.
$$

En este punto se ha completado únicamente la **eliminación hacia adelante**. La solución del sistema requiere posteriormente realizar sustitución hacia atrás.

Después se ejecuta el algoritmo deseado.
