# Diapositiva 2. Representación posicional

Recordemos

$
x=\sum_{k=-m}^{n} a_k\beta^k.
$

Comprobemos esta representación con un ejemplo decimal.


```python
digits=[2,5,4,7,3]
powers=[2,1,0,-1,-2]
value=sum(d*10**p for d,p in zip(digits,powers))
for d,p in zip(digits,powers):
    print(f"{d}·10^{p} = {d*10**p}")
print("\nValor =",value)
```

# Diapositiva 3. Representación de punto flotante

Un número puede escribirse como

$
x=f\beta^E.
$

En base 10 esto corresponde a la notación científica.


```python
nums=[347.26,0.0034726,125000]
for x in nums:
    print(f"{x:.6e}")
```

Observe cómo Python expresa automáticamente los números en notación científica.

# Diapositiva 4. Representación normalizada

Una mantisa normalizada satisface

$
1\le f<\beta.
$


Para base 10:


```python
nums=[347.26,0.0034726,125000]
for x in nums:
    import math
    if x!=0:
        E=math.floor(math.log10(abs(x)))
        f=x/10**E
        print(f"x={x}, f={f}, E={E}")
```

Verifique que en todos los casos la mantisa cumple $1\le f<10$.

# Diapositiva 6. Bit implícito

En IEEE‑754 el primer bit de la mantisa normalizada siempre es 1 y no se almacena explícitamente.

Visualicemos los bits de un número real.


```python
import struct
x=3.141592653589793
bits=''.join(f'{b:08b}' for b in struct.pack('>d',x))
print(bits)
print("\nSigno     :",bits[0])
print("Exponente :",bits[1:12])
print("Mantisa   :",bits[12:])
```

Localice el bit de signo, el exponente y la mantisa.
