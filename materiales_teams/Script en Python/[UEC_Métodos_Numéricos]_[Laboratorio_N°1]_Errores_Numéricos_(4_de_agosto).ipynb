# 1. Representación de números en punto flotante

En el sistema decimal, algunos números poseen una representación finita. Sin embargo, al convertirlos a binario, su representación puede ser infinita.

Por esta razón, Python almacena una aproximación del número.

````python
x = 0.1
y = 0.2

print("x =", x)
print("y =", y)
print("x + y =", x + y)
print("¿x + y es exactamente 0.3?", x + y == 0.3)
print(f"x + y con 20 cifras decimales: {x + y:.20f}")
````

### Análisis guiado

Observe que la expresión `0.1 + 0.2 == 0.3` produce `False`.

Esto no significa que Python realice incorrectamente la suma. El resultado aparece porque `0.1`, `0.2` y `0.3` no tienen una representación binaria finita exacta.

# 2. Error absoluto, relativo y porcentual

Sea \(x\) un valor exacto y $\widetilde{x}$ una aproximación.

El error absoluto se define como

$
E_a=|x-\widetilde{x}|.
$

El error relativo se define como

$
E_r=\frac{|x-\widetilde{x}|}{|x|},
\qquad x \neq 0.
$

El error porcentual se define como

$
E_p=100E_r.
$

````python
import math

valor_exacto = math.pi
valor_aproximado = 3.14

error_absoluto = abs(valor_exacto - valor_aproximado)
error_relativo = error_absoluto / abs(valor_exacto)
error_porcentual = 100 * error_relativo

print("Valor exacto:", valor_exacto)
print("Valor aproximado:", valor_aproximado)
print("Error absoluto:", error_absoluto)
print("Error relativo:", error_relativo)
print("Error porcentual:", error_porcentual, "%")
````

# 3. Redondeo y truncamiento

El redondeo y el truncamiento reducen el número de cifras de una cantidad, pero no operan de la misma manera.

En el truncamiento se eliminan las cifras posteriores a una posición determinada.

En el redondeo se observa la primera cifra descartada para decidir si la última cifra conservada permanece igual o aumenta en una unidad.

````python
# ==========================================================
# Redondeo y truncamiento
# ==========================================================

# Importa la librería matemática
import math

# ----------------------------------------------------------
# Función para truncar un número
# ----------------------------------------------------------

# Esta función recibe un número y la cantidad de decimales
# que se desean conservar.
def truncar(x, decimales):

    # Calcula el factor de escala correspondiente
    factor = 10 ** decimales

    # Elimina los decimales adicionales y devuelve el número truncado
    return math.trunc(x * factor) / factor

# ----------------------------------------------------------
# Número a aproximar
# ----------------------------------------------------------

# Se toma el valor de π como ejemplo
numero = math.pi

# ----------------------------------------------------------
# Cálculo del redondeo y el truncamiento
# ----------------------------------------------------------

# Recorre el número de decimales desde 1 hasta 5
for n in range(1, 6):

    # Calcula el número redondeado a n decimales
    redondeado = round(numero, n)

    # Calcula el número truncado a n decimales
    truncado = truncar(numero, n)

    # Muestra los resultados obtenidos
    print(
        f"Decimales: {n} | "
        f"Redondeado: {redondeado} | "
        f"Truncado: {truncado}"
    )
````

# 4. Épsilon de máquina

En un computador, los números reales no pueden almacenarse con precisión infinita. Debido a ello, existe un límite a partir del cual un número positivo muy pequeño deja de producir cambios cuando se suma a otro número.

Por ejemplo, considere el número $25.$

Si se suma una cantidad relativamente grande, como $25+0.001,$ el resultado cambia y el computador puede representarlo sin dificultad.

Sin embargo, si se suma un número extremadamente pequeño, por ejemplo $25+0.000000000000000000000000001,$ el computador puede almacenar el resultado simplemente como $25,$ porque esa diferencia es demasiado pequeña para ser representada con la precisión disponible.

El **épsilon de máquina**, denotado por $\varepsilon$, es el menor número positivo que, al sumarse a $1$, produce un resultado diferente de $1$ en la aritmética del computador.

Matemáticamente,

$
1+\varepsilon>1.
$

El valor de $\varepsilon$ depende del tipo de dato utilizado y representa una medida de la precisión con la que el computador almacena los números reales.

````python
# ==========================================================
# ¿Qué ocurre al sumar números muy pequeños a 25?
# ==========================================================

# Número de referencia
x = 25.0

# Valor que se irá haciendo cada vez más pequeño
incremento = 1.0

print("Número de referencia:", x)
print()

# Se repite el proceso hasta que el computador ya no detecte cambios
while x + incremento != x:

    print("Incremento =", incremento)
    print("x + incremento =", x + incremento)
    print()

    # El incremento se divide entre 2
    incremento = incremento / 2

print("El computador ya no puede distinguir entre:")
print(f"{x} y {x + incremento}")

print("\nÚltimo incremento que produjo un cambio:")

# El último incremento útil es el doble del último valor
print(2 * incremento)
````

# 5. Cancelación numérica

En algunas operaciones aritméticas, el computador debe restar dos números muy cercanos entre sí. Cuando esto ocurre, muchas de las cifras significativas coinciden y se cancelan durante la resta.

Como consecuencia, el resultado puede perder precisión y contener un error relativo considerable, aun cuando los números originales hayan sido representados con buena precisión. Este fenómeno se conoce como **cancelación numérica** o **pérdida de cifras significativas**.

Considere la función

$$
f(x)=\sqrt{x+1}-\sqrt{x}.
$$

Si \(x\) es pequeño, ambas raíces son suficientemente diferentes y la resta puede realizarse sin mayores inconvenientes.

Sin embargo, cuando \(x\) es muy grande, los valores de

$$
\sqrt{x+1}
\qquad \text{y} \qquad
\sqrt{x}
$$

son extremadamente cercanos. En consecuencia, la resta puede perder precisión debido a la cancelación de cifras significativas.

Por ejemplo, si

$$
x=10^{12},
$$

entonces

$$
\sqrt{x+1}\approx 1000000.0000005,
$$

mientras que

$$
\sqrt{x}=1000000.
$$

Aunque ambos números son grandes, su diferencia es muy pequeña. En este caso, los errores de redondeo propios de la representación en punto flotante pueden afectar significativamente el resultado de la resta.

Una forma de evitar este problema consiste en racionalizar la expresión.

Multiplicando y dividiendo por el conjugado, se obtiene

$$
\sqrt{x+1}-\sqrt{x}
=
\frac{1}{\sqrt{x+1}+\sqrt{x}}.
$$

Las dos expresiones son **algebraicamente equivalentes**, es decir, producen el mismo resultado exacto desde el punto de vista matemático. Sin embargo, desde el punto de vista numérico, la segunda expresión suele ser mucho más estable, ya que evita la resta de cantidades muy próximas entre sí.

En el siguiente ejemplo se compararán ambas expresiones para observar cómo cambia el error cuando el valor de \(x\) aumenta.

````python
# ==========================================================
# Cancelación numérica
# ==========================================================

# Importa la librería matemática
import math

# ----------------------------------------------------------
# Forma directa
# ----------------------------------------------------------

# Calcula la expresión original
#
#          √(x+1) - √x
#
# Cuando x es muy grande, la resta entre dos números muy
# cercanos puede producir pérdida de cifras significativas.
def forma_directa(x):
    return math.sqrt(x + 1) - math.sqrt(x)

# ----------------------------------------------------------
# Forma racionalizada
# ----------------------------------------------------------

# Calcula la expresión racionalizada
#
#               1
# -------------------------------
#   √(x+1) + √x
#
# Esta forma evita la resta entre cantidades muy próximas
# y, por tanto, suele ser más estable numéricamente.
def forma_estable(x):
    return 1 / (math.sqrt(x + 1) + math.sqrt(x))

# ----------------------------------------------------------
# Valores de prueba
# ----------------------------------------------------------

# Lista de valores para comparar ambas expresiones
valores = [10, 10**4, 10**8, 10**12, 10**16]

# Imprime el encabezado de la tabla
print("=" * 120)

print(f"{'x':>15}"
      f"{'Forma directa':>28}"
      f"{'Forma estable':>28}"
      f"{'Error relativo':>25}")

print("-" * 120)

# ----------------------------------------------------------
# Comparación
# ----------------------------------------------------------

# Recorre todos los valores de la lista
for x in valores:

    # Calcula la forma directa
    directa = forma_directa(x)

    # Calcula la forma racionalizada
    estable = forma_estable(x)

    # Calcula el error relativo tomando como referencia
    # la forma racionalizada
    error_relativo = abs(directa - estable) / abs(estable)

    # Imprime los resultados
    print(f"{x:>15}"
          f"{directa:>28.18e}"
          f"{estable:>28.18e}"
          f"{error_relativo:>25.10e}")

print("=" * 120)
````

# 6. Precisión simple y doble precisión

En Python, la biblioteca **NumPy** permite representar números reales utilizando diferentes niveles de precisión.

Los tipos de dato más utilizados son:

- `float32`: utiliza **32 bits** para almacenar un número real (precisión simple).
- `float64`: utiliza **64 bits** para almacenar un número real (doble precisión).

En general, un número almacenado como `float64` conserva un mayor número de cifras significativas que uno almacenado como `float32`.

Para ilustrar esta diferencia, considere el número

$$
\frac{1}{3}=0.333333333333\ldots
$$

Como su expansión decimal es infinita, el computador solo puede almacenar una aproximación.

Al representar este número con `float32` y `float64`, se observará que la aproximación en doble precisión conserva un mayor número de cifras correctas.

En el siguiente ejemplo se compararán ambas representaciones mostrando sus valores con veinte cifras decimales.
